# Affordable contract picker
Use the rule-based contract picker to propose 1-5 candidates per ticker and horizon within a fixed budget.

```python
%pip -q install -r ../requirements.txt
from quantflow.recommend.options_picker import pick_affordable_contracts
tickers = ['AAPL','MSFT']
for t in tickers:
    for h in ['1w','1m','3m']:
        ideas = pick_affordable_contracts(t, h, bias='long', budget=300)
        print(f"\n{t} {h} long candidates:")
        for i in ideas:
            print(i)
```
