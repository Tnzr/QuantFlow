# Rule-based recommender
Generate horizon-based recommendations without LLM.

```python
%pip -q install -r ../requirements.txt
from quantflow.recommend.engine import RuleEngine
from quantflow.data.universe import HIGH_INTEREST
from quantflow.data.persistence import save_recommendations, create_schema

# ensure DB schema
create_schema()

eng = RuleEngine()
TICKERS = ['AAPL','MSFT','NVDA','AMZN']
recs_all = []
for t in TICKERS:
    recs = eng.recommend(t)
    for r in recs:
        print(f"{t} {r.horizon} {r.bias} conf={r.confidence:.2f} entry={r.entry:.2f} stop={r.stop:.2f} target={r.target:.2f} | {r.notes}")
    recs_all.extend(recs)

# persist
save = True
if save:
    save_recommendations(recs_all)
    print("Saved recommendations to DB")
```
