# Option chain exploration and filters
Load option chains with yfinance, apply liquidity and budget filters, and inspect candidates.

```python
%pip -q install -r ../requirements.txt
import pandas as pd
from quantflow.options.chain import load_chain
ticker = 'AAPL'
df = load_chain(ticker)
df.head()
```

```python
budget = 300.0
max_spread_abs = 0.10
max_spread_pct = 0.08
min_oi = 200
min_vol = 50

df = df.copy()
df['mid'] = (df['bid'].fillna(0)+df['ask'].fillna(0))/2
df = df[(df['ask']>0)&(df['mid']>0)&(df['ask']<=budget)]
df = df[(df['open_interest']>=min_oi)&(df['volume']>=min_vol)]
df['spread']= (df['ask']-df['bid']).clip(lower=0)
df = df[(df['spread']<=max_spread_abs)|(df['spread']/df['mid']<=max_spread_pct)]
df.sort_values(['expiry','right','strike']).head(20)
```
