# Finviz scan and cache
This notebook runs predefined Finviz presets, caches results to SQLite, and previews latest snapshots.

```python
%pip -q install -r ../requirements.txt
from quantflow.data.finviz_client import PRESETS, run_screener
from quantflow.data.persistence import create_schema, save_finviz_snapshot
from quantflow.data import queries

create_schema()
for name in PRESETS.keys():
    print(f"Running preset: {name}")
    df = run_screener(name)
    print(df.head(3))
    save_finviz_snapshot(df, name)
print("Saved snapshots.")
```

```python
from quantflow.data.queries import latest_by_preset
for name in PRESETS.keys():
    latest = latest_by_preset(name)
    print(name, latest.shape)
    display(latest.head(10))
```
