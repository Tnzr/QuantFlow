# Seasonality and indicators demo
Compute DOY seasonality and standard indicators for a ticker.

```python
%pip -q install -r ../requirements.txt
from quantflow.features.seasonality import seasonality_by_doy
from quantflow.features.indicators import fetch_ohlcv, compute_indicators
import plotly.express as px

ticker = 'AAPL'
doy = seasonality_by_doy(ticker, years=10)
fig = px.line(doy, x='doy', y='avg_ret', title=f'{ticker} seasonality avg daily return by DOY')
fig.show()

ohlcv = fetch_ohlcv(ticker)
ind = compute_indicators(ohlcv)
ind.tail()
```
