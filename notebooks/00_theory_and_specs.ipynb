# Engineering specs and plan (PoC/MVP)

## Goals
- Risk-first options workflow for low capital ($100–300).
- Automated scouting via Finviz, rule-based recommender (no LLM), later add sentiment agents.
- Persistent cache (SQLite) to build a RAG dataset over time.

## Golden rules
- Risk ≤1% per trade; prefer defined-risk spreads.
- Pre-commit time stops (avoid holding weeklies into last 2 days).
- ATR-based stops; avoid earnings for short-dated unless IV play.
- Liquidity filters on chains (tight spreads, OI/Vol).

## Data
- Finviz screener presets; hourly/daily snapshots.
- yfinance OHLCV for indicators, seasonality, earnings hints.
- Option chains for affordability/liquidity ranking.

## Features
- Indicators: RSI, ATR, SMA20/50/200, vol.
- Seasonality: DOY averages; earnings proximity.
- Recommender: horizon-based rules + ATR stops/targets.

## Roadmap
1) Screeners + cache + CLI.
2) Rule engine + option picker.
3) Backtests + UI (Streamlit).
4) Sentiment/RAG agents.
